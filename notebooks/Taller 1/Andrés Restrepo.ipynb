# Taller análisis descriptivo de variables cualitativas

Se va a trabajar con un dataset que contiene información sobre incidentes de tránsito en Medellín con víctimas.

Cárguelo y verifique que se haya cargado correctamente.

```python
import pandas as pd

df_accidentes=pd.read_csv("/content/sample_data/Accidentes de tránsito en Medellín.csv")
df_accidentes.head(5)
```

La variable "Fecha_incidente" debe convertirse a dato de tiempo.

Excluya los datos del año 2021.

```python
df_accidentes['Fecha_incidente']=pd.to_datetime(df_accidentes['Fecha_incidente'])
df_accidentes=df_accidentes[df_accidentes['Año']!='2021']
df_accidentes.head()
```

De acuerdo a lo visto en clase, indique qué tipo de variable es cada una de las que están en el DataFrame.

Convierta a "category" las variables categóricas, cuidando que las variables nominales y ordinales queden correctamente convertidas.

 #   Tipos de variables
---  ------            --------------   -----         
 0   Fecha_incidente  - Cuantitativa intervalo
 1   Gravedad_victima - Cualitativa nominal      
 2   Clase_incidente  - Cualitativa nominal      
 3   Sexo             - Cualitativa nominal      
 4   Condicion        - Cualitativa nominal       
 5   Grupo_edad       - Cuantitativa intervalo      
 6   Comuna           - Cualitativa nominal         
 7   Barrio           - Cualitativa nominal       
 8   Dia_semana       - Cuantitativa ordinal      
 9   Mes              - Cuantitativa ordinal       
 10  Año              - Cuantitativa ordinal    

```python
##Obetener información de los tipos de datos de cada una de las columnas
df_accidentes.info()

##Seleccionar unicamente variables categoricas
df_accidentes=df_accidentes.select_dtypes("object")
df_accidentes.info

```

```python
##Convertir a category
df_accidentes['Gravedad_victima']=df_accidentes['Gravedad_victima'].astype('category')
df_accidentes['Clase_incidente']=df_accidentes['Clase_incidente'].astype('category')
df_accidentes['Sexo']=df_accidentes['Sexo'].astype('category')
df_accidentes['Condicion']=df_accidentes['Condicion'].astype('category')
df_accidentes['Grupo_edad']=df_accidentes['Grupo_edad'].astype('category')
df_accidentes['Comuna']=df_accidentes['Comuna'].astype('category')
df_accidentes['Barrio']=df_accidentes['Barrio'].astype('category')

#Crear categorías

df_accidentes['Grupo_edad']=pd.Categorical(
    df_accidentes['Grupo_edad'],
    categories= ['0 - 9', '10 - 19', '20 - 29', '30 - 39','40 - 49', '50 - 59', '60 - 69',
                         '70 - 79', '80 o más'],
    ordered=True
)

df_accidentes['Dia_semana']=pd.Categorical(
    df_accidentes['Dia_semana'],
    categories=['Lunes', 'Martes', 'Miércoles', 'Jueves','Viernes', 'Sábado', 'Domingo'],
    ordered=True
)

df_accidentes['Mes']=pd.Categorical(
    df_accidentes['Mes'],
    categories=['Enero', 'Febrero', 'Marzo', 'Abril', 'Mayo', 'Junio', 'Julio',
       'Agosto', 'Septiembre', 'Octubre', 'Noviembre', 'Diciembre'],
    ordered=True
)


df_accidentes.info()
```

Para las variables ordinales, calcule la mediana, y los quartiles 1, 2 y 3.

```python
df_accidentes['Grupo_edad'].value_counts(normalize=False, sort=False).cumsum().plot(kind='bar')
```

```python
print("La mediana es: ", df_accidentes['Grupo_edad'].cat.categories[df_accidentes['Grupo_edad'].cat.codes.median().astype(int)])
print("El quartil 1 es: ", df_accidentes['Grupo_edad'].cat.categories[df_accidentes['Grupo_edad'].cat.codes.quantile([0.25]).astype(int)])
print("El quartil 2 es: ", df_accidentes['Grupo_edad'].cat.categories[df_accidentes['Grupo_edad'].cat.codes.quantile([0.50]).astype(int)])
print("El quartil 3 es: ", df_accidentes['Grupo_edad'].cat.categories[df_accidentes['Grupo_edad'].cat.codes.quantile([0.75]).astype(int)])
```
