Para las variables nominales, haga un gráfico de barras horizontales de la tabla de frecuencia relativa, ordenados de forma descendente.

Para las variables ordinales, haga un gráfico de barras verticales de la tabla de frecuencia relativa, ordenados de acuerdo al orden de las categorías.

```python

freq_rel = df["Clase_incidente"].value_counts(normalize=True) * 100
freq_rel.sort_values(ascending=True).plot(
    kind="barh", color="skyblue", xlabel="Porcentaje (%)", title="Clase de Incidente"
)
```

```python
freq_rel = df["Sexo"].value_counts(normalize=True) * 100
freq_rel.sort_values(ascending=True).plot(
    kind="barh", color="red", xlabel="Porcentaje (%)", title="Sexo"
)
```

```python
freq_rel = df["Condicion"].value_counts(normalize=True) * 100
freq_rel.sort_values(ascending=True).plot(
    kind="barh", color="Purple", xlabel="Porcentaje (%)", title="Condición"
)
```

```python

freq_rel_ordinal = (
    df["Gravedad_victima"].value_counts(normalize=False) * 100)
freq_rel_ordinal.plot(
    kind="bar", color="orange", ylabel="Porcentaje (%)", title="Gravedad de la Víctima")
```

```python
Orden = ["Lunes", "Martes", "Miercoles", "Jueves", "Viernes", "Sabado", "Domingo"]
freq_rel_ordinal = (
    df["Dia_semana"].value_counts(normalize=True, sort = True)* 100)
freq_rel_ordinal.plot(
    kind="bar", color="Green", ylabel="Porcentaje (%)", title="Día de la semana")
```

```python

freq_rel_ordinal = (
    df["Mes"].value_counts(normalize=False, sort =True) * 100)
freq_rel_ordinal.plot(
    kind="bar", color="Purple", ylabel="Porcentaje (%)", title="Mes")
```

```python
freq_rel_ordinal = (
    df["Año"].value_counts(normalize=False) * 100)
freq_rel_ordinal.plot(
    kind="bar", color="Blue", ylabel="Porcentaje (%)", title="Año")
```
