# Taller análisis descriptivo de variables cualitativas

Se va a trabajar con un dataset que contiene información sobre incidentes de tránsito en Medellín con víctimas.

Cárguelo y verifique que se haya cargado correctamente.

```python
import pandas as pd

df = pd.read_csv("Accidentes de tránsito en Medellín.csv")

df.head()
```

La variable "Fecha_incidente" debe convertirse a dato de tiempo.

Excluya los datos del año 2021.

```python
df["Fecha_incidente"] = pd.to_datetime(df["Fecha_incidente"])

df = df[df["Fecha_incidente"].dt.year != 2021]

df.head()
```

De acuerdo a lo visto en clase, indique qué tipo de variable es cada una de las que están en el DataFrame.

Convierta a "category" las variables categóricas, cuidando que las variables nominales y ordinales queden correctamente convertidas.

Las variables son:

* Fecha_incidente - Cuantitativa de intervalo
* Gravedad_victima - Cualitativa ordinal
* Clase_incidente - Cualitativa nominal
* Sexo - Cualitativa nominal
* Condicion - Cualitativa nominal
* Grupo_edad - Cualitativa ordinal
* Comuna - Cualitativa nominal
* Barrio - Cualitativa nominal
* Dia_semana - Cualitativa ordinal
* Mes - Cualitativa ordinal
* Año - Cuantitativa discreta de intervalo

```python
df1 = df.copy()
```

```python
df1['Clase_incidente'] = df1['Clase_incidente'].astype('category')
df1['Sexo'] = df1['Sexo'].astype('category')
df1['Condicion'] = df1['Condicion'].astype('category')
df1['Comuna'] = df1['Comuna'].astype('category')
df1['Barrio'] = df1['Barrio'].astype('category')

df1['Gravedad_victima'] = pd.Categorical(
    df1['Gravedad_victima'],
    categories=['Heridos', 'Muertos'],
    ordered=True
)

df1['Grupo_edad'] = pd.Categorical(
    df1['Grupo_edad'],
    categories=[
        '0 - 9', '10 - 19', '20 - 29', '30 - 39',
        '40 - 49', '50 - 59', '60 - 69',
        '70 - 79', '80 o mas'
    ],
    ordered=True
)

df1['Dia_semana'] = pd.Categorical(
    df1['Dia_semana'],
    categories=[
        'Lunes', 'Martes', 'Miércoles',
        'Jueves', 'Viernes', 'Sábado', 'Domingo'
    ],
    ordered=True
)

df1['Mes'] = pd.Categorical(
    df1['Mes'],
    categories=[
        'Enero', 'Febrero', 'Marzo', 'Abril',
        'Mayo', 'Junio', 'Julio', 'Agosto',
        'Septiembre', 'Octubre', 'Noviembre', 'Diciembre'
    ],
    ordered=True
)
df1.head()
```

Haga las tablas de frecuencias porcentuales de todas las variables categóricas.

Para cada variable, indique cuál es la categoría más frecuente y cuál la menos frecuente.

Identifique qué variables presentan categorías atípicas. Indique los nombres de las categorías atípicas.

```python
columnas_categoricas = df1.select_dtypes(include='category').columns

for col in columnas_categoricas:
    frec = df1[col].value_counts(normalize=True) * 100
    print(f'\n{"="*50}')
    print(frec)
    print('\nCategoría más frecuente:')
    print(f'{frec.idxmax()} ({frec.max():.2f}%)')
    print('\nCategoría menos frecuente:')
    print(f'{frec.idxmin()} ({frec.min():.2f}%)')
    print('\nCategorías atípicas (<1%):')
    print(frec[frec < 1])
```

Gravedad_victima:
La categoría más frecuente es Heridos con 99.17% de los registros. La categoría menos frecuente es Muertos con 0.83%. Dado que Muertos representa menos del 1% de los casos, se considera una categoría atípica.

Clase_incidente:
La categoría más frecuente es Choque con 43.97% de los registros. La categoría menos frecuente es Incendio con 0.01%. La categoría Incendio puede considerarse atípica debido a su muy baja frecuencia.

Sexo:

Grupo_edad:
