La variable "Fecha_incidente" debe convertirse a dato de tiempo.

Excluya los datos del año 2021.

```python
df['Fecha_incidente'] = pd.to_datetime(df['Fecha_incidente']) # Convertir en fecha

df2 = df[df['Fecha_incidente'].dt.year != 2021] # sIN 2021
df2.head()

```

**Convierta a "category" las variables categóricas, cuidando que las variables nominales y ordinales queden correctamente convertidas.**

**RESPUESTA: **
*Se cambia a "category" las que sean tipo "object" y sean nomimales.*


```python
df3 = df2.copy()

df4 = ['Gravedad_victima', 'Clase_incidente', 'Sexo', 'Condicion', 'Grupo_edad', 'Comuna', 'Barrio']
df3[df4] = df3[df4].astype('category') # Variables nominales convertidas a tipo category
df3.info()
```

```python
# Variables ordinales convertidas a tipo category
df3['Grupo_edad'] = pd.Categorical(
    df3['Grupo_edad'],
    categories=['0 - 9', '10 - 19', '20 - 29', '30 - 39', '40 - 49', '50 - 59',
                '60 - 69', '70 - 79', '80 o más'],
    ordered=True # Variable ordenada
    )

df3.info()
```

```python
# Variables ordinales convertidas a tipo category
df3['Mes'] = pd.Categorical(
    df3['Mes'],
    categories=['Enero', 'Febrero', 'Marzo', 'Abril', 'Mayo', 'Junio',
                'Julio', 'Agosto', 'Septiembre', 'Octubre', 'Noviembre',
                ' Diciembre'],
    ordered=True # Variable ordenada
    )

df3.info()
```

```python
# Variables ordinales convertidas a tipo category
df3['Dia_semana'] = pd.Categorical(
    df3['Dia_semana'],
    categories=['Lunes', 'Martes', 'Miercoles', 'Jueves', 'Viernes', 'Sabado',
                'Domingo'],
    ordered=True # Variable ordenada
    )

df3.info()
```
