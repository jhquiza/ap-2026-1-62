La variable "Fecha_incidente" debe convertirse a dato de tiempo.

Excluya los datos del año 2021.

```python
df['Fecha_incidente'] = pd.to_datetime( df['Fecha_incidente'],)
df_año_2021=df[df['Fecha_incidente'].dt.year!=2021]
df_año_2021.head()
```

De acuerdo a lo visto en clase, indique qué tipo de variable es cada una de las que están en el DataFrame.

Convierta a "category" las variables categóricas, cuidando que las variables nominales y ordinales queden correctamente convertidas.

1. Fecha_incidente:Cualitativa_nominal
2. Gravedad_victima: Cualitativa_ nomimal    
3. Clase:incidente:Cualitativa_nominal
4. Sexo: Cualitativa_nominal
5. Condicion: Cualitativa_ nomimal               
6. Grupo_edad:cuantitativa_nomimal            
7. Comuna:Cualitativa_nominal                 
8. Barrio:Cualitativa_nominal

10. Mes:Cualitativa_ordinales                     
11. Año:cuantitativa_continua_nominal

```python
df1=df_año_2021.copy()
df1.info()
```

```python
df1['Fecha_incidente'] = df1['Fecha_incidente'].astype('category')
df1['Gravedad_victima'] = df1['Gravedad_victima'].astype('category')
df1['Clase_incidente'] = df1['Clase_incidente'].astype('category')
df1['Sexo'] = df1['Sexo'].astype('category')
df1['Condicion'] = df1['Condicion'].astype('category')
df1['Grupo_edad'] = df1['Grupo_edad'].astype('category')
df1['Barrio'] = df1['Barrio'].astype('category')
df1['Comuna'] = df1['Comuna'].astype('category')
df1['Año'] = df1['Año'].astype('category')

df1.info()
```

```python
df1['Dia_semana'] = pd.Categorical(
    df1['Dia_semana'],
    categories=['Lunes','Martes','Miércoles','Jueves','Viernes','Sábado','Domingo'],
    ordered=True # variable ordenada si dice false es que no la organice
    )

df1.info()
```

```python
df1['Mes'] = pd.Categorical(
    df1['Mes'],
    categories=['Enero','Febrero','Marzo','Abril','Mayo','Junio','Julio','Agosto','Septiembre','Octubre','Noviembre','Diciembre'],
    ordered=True # variable ordenada si dice false es que no la organice
    )

df1.info()
```
