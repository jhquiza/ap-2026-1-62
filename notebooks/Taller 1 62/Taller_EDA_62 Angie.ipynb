## Paso 4: Análisis de variables cuantitativas

Haga análisis de normalidad de las variables cuantitativas. Indique cuáles variables cumplen con el supuesto de normalidad y cuáles no.

Haga un análisis gráfico de outliers, y como resultado de este indique que variables tienen datos atípicos.

```python
variables_numericas = df.select_dtypes(include=['int64', 'float64']).columns
variables_numericas
```

```python
from scipy.stats import shapiro
```

```python
for columna in variables_numericas:
    stat, p = shapiro(df[columna].dropna())

    print("Variable:", columna)
    print("p-value:", p)

    if p > 0.05:
        print("La variable tiene distribución normal")
    else:
        print("La variable no tiene distribución normal")

    print("-----------------------")
```

```python
import matplotlib.pyplot as plt
```

```python
for columna in variables_numericas:
    plt.figure(figsize=(6,3))
    plt.boxplot(df[columna].dropna())
    plt.title("Boxplot de " + columna)
    plt.ylabel(columna)
    plt.show()
```

## Respuestas paso 4

**Análisis de normalidad**
Los resultados obtenidos muestran valores de p-value menores a 0.05 en todas las variables por lo tanto se rechaza la hipótesis de normalidad y ninguna de las variables cuantitativas cumple con el supuesto de distribución normal

**Análisis gráfico de outliers**
Se realizaron diagramas de caja y bigotes para identificar posibles valores atípicos en las variables cuantitativas y se identificaron datos atípicos en las variables: A1, A2, A7, A10, A13 y A14 estas variables presentan valores alejados de la distribución principal de los datos, representados mediante puntos fuera de los límites del boxplot

## Paso 6: Análisis bivariado

Se va a suponer que con este archivo de datos se desea hacer un modelo de aprendizaje de máquina que prediga la vaariable **A15**.


Mediante análisis de gráficas responda las siguientes preguntas:

- ¿Cómo se relaciona cada una de las variables cuantitativas con la variable que se desea predecir?
- ¿Cómo se relaciona cada una de las variables cualitativas con la variable que se desea predecir?
- De acuerdo con estos análisis, ¿cuáles variables parecerían ser buenas predictoras, y cuáles no?


```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
variables_numericas = ['A1','A2','A7','A10','A13','A14']

for columna in variables_numericas:
    plt.figure(figsize=(6,4))

    sns.boxplot(
        data=df,
        x='A15',
        y=columna
    )

    plt.title(f'{columna} vs A15')
    plt.xlabel('Aprobación (A15)')
    plt.ylabel(columna)
    plt.show()
```

```python
for columna in variables_categoricas:

    tabla = pd.crosstab(
        df[columna],
        df['A15']
    )

    tabla.plot(
        kind='bar',
        figsize=(7,4)
    )

    plt.title(f'{columna} vs A15')
    plt.xlabel(columna)
    plt.ylabel('Cantidad')
    plt.xticks(rotation=45)

    plt.show()
```

```python
for columna in variables_categoricas:

    print("Variable:", columna)

    porcentaje = pd.crosstab(
        df[columna],
        df['A15'],
        normalize='index'
    ) * 100

    print(porcentaje.round(2))

    print("----------------------------")
```

# Respuestas paso 6

**¿Cómo se relaciona cada una de las variables cuantitativas con la variable que se desea predecir?**
Al comparar las distribuciones de las variables numéricas entre las categorías, se observaron diferencias en algunas variables, variables como A1, A2 y A7 presentan cambios en sus valores centrales entre las clases de aprobación y rechazo por lo que podrían tener relación con la variable objetivo. Además algunas variables como A10, A13 y A14 presentan dispersión y valores atípicos pero también muestran diferencias entre grupos por lo que podrían aportar información al modelo predictivo

**¿Cómo se relaciona cada una de las variables cualitativas con la variable que se desea predecir?**
Se identificaron diferencias entre las categorías de algunas variables, principalmente las variables A8 y A9 presentan una relación más evidente con A15, debido a que algunas de sus categorías tienen una mayor proporción de aprobaciones o rechazos. Las variables A5 y A6 también muestran diferencias entre sus categorías indicando que pueden contener información relevante para predecir la aprobación del crédito. Por otro lado, variables como A0, A3, A4, A11 y A12 presentan diferencias menos marcadas entre sus categorías, por lo que aparentemente tienen menor relación con la variable objetivo

**¿Cuáles variables parecerían ser buenas predictoras y cuáles no?**
De acuerdo con el análisis gráfico realizado las variables que parecen tener mayor capacidad predictiva para A15 son: A8, A9, A5, A6, A1 y A2 y las variables A0, A3, A4, A11 y A12 parecen tener una relación más débil con la variable objetivo
