## Paso 4: Análisis de variables cuantitativas

Haga análisis de normalidad de las variables cuantitativas. Indique cuáles variables cumplen con el supuesto de normalidad y cuáles no.

Haga un análisis gráfico de outliers, y como resultado de este indique que variables tienen datos atípicos.

```python
num_vars = df.select_dtypes(include=["number"]).columns
print('Variables numéricas: ', num_vars)

df.hist(figsize=(16, 8), bins=20, layout=(2,3), grid=False, sharex=False, sharey=False);
```

```python
from scipy.stats import kstest

for num in num_vars:
    __, p_value = kstest(df[num], 'norm', args=(df[num].mean(), df[num].std()))
    print(f"Prueba de Kolmogorov-Smirnov para {num}:")
    print(f"  Valor p: {p_value}")
    if p_value > 0.05:
        print(f"  {num} parece ser normal (no se rechaza la hipótesis nula)")
    else:
        print(f"  {num} no parece ser normal (se rechaza la hipótesis nula)")
    print()
```

```python
import matplotlib.pyplot as plt

for num in num_vars:
    plt.figure()
    df.boxplot(column=num, grid=False, vert=False)
    plt.title(f'Boxplot de {num}')
    plt.show()
```



Basado en la prueba de Kolmogorov-Smirnov aplicada a las variables numéricas (`A1`, `A2`, `A7`, `A10`, `A13`, `A14`):

*   **Ninguna de las variables cuantitativas cumple con el supuesto de normalidad.**
*   Para `A2`, `A7`, `A10`, `A14`, los valores p fueron extremadamente bajos (mucho menores a 0.05), lo que lleva a rechazar la hipótesis nula de normalidad. Esto significa que estas variables no siguen una distribución normal.
*   Para `A1` y `A13`, la prueba arrojó un valor p de `nan`. Esto es probable debido a la presencia de valores nulos en estas columnas. Sin embargo, en el contexto de las demás variables y la inspección visual de los histogramas (previamente generados), es muy improbable que estas variables sean normales.


**Análisis Gráfico de Outliers (Valores Atípicos):**

Al revisar los boxplots de las variables cuantitativas pudimos observar lo siguiente:

*   **Todas las variables cuantitativas (`A1`, `A2`, `A7`, `A10`, `A13`, `A14`) presentan datos atípicos.**
*   Esto se evidencia por los puntos individuales que se extienden significativamente más allá de los bigotes de los diagramas de caja.
*   Particularmente, la variable **A14** muestra una gran cantidad de outliers, algunos de ellos con valores muy elevados, lo que ya había sido sugerido por las estadísticas descriptivas. Otras variables como `A1`, `A2`, `A7`, `A10`, y `A13` también exhiben valores atípicos, aunque quizás en menor medida o con menor dispersión que `A14`.

## Paso 6: Análisis bivariado

Se va a suponer que con este archivo de datos se desea hacer un modelo de aprendizaje de máquina que prediga la vaariable **A15**.


Mediante análisis de gráficas responda las siguientes preguntas:

- ¿Cómo se relaciona cada una de las variables cuantitativas con la variable que se desea predecir?
- ¿Cómo se relaciona cada una de las variables cualitativas con la variable que se desea predecir?
- De acuerdo con estos análisis, ¿cuáles variables parecerían ser buenas predictoras, y cuáles no?


```python
import seaborn as sns

quantitative_vars = ['A1', 'A2', 'A7', 'A10', 'A13', 'A14']
target_var = 'A15'

for var in quantitative_vars:
    plt.figure(figsize=(8, 6))
    sns.boxplot(x=target_var, y=var, data=df)
    plt.title(f'Distribution of {var} by {target_var}')
    plt.xlabel(target_var)
    plt.ylabel(var)
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

qualitative_vars = ['A0', 'A3', 'A4', 'A5', 'A6', 'A8', 'A9', 'A11', 'A12']
target_var = 'A15'

for var in qualitative_vars:
    plt.figure(figsize=(10, 6))


    crosstab_df = pd.crosstab(df[var], df[target_var], normalize='index')


    crosstab_df.plot(kind='bar', stacked=True, figsize=(10, 6))
    plt.title(f'Proportion of {target_var} by {var}')
    plt.xlabel(var)
    plt.ylabel(f'Proportion of {target_var}')
    plt.xticks(rotation=45, ha='right')
    plt.legend(title=target_var)
    plt.tight_layout()
    plt.show()
```

### Análisis respecto a A15

1. - **A1 y A13:** presentan distribuciones similares entre `+` y `-`, por lo que tienen poca relación con A15 y no son fuertes predictorias de la misma.
- **A2, A7, A10 y A14:** muestran diferencias en sus medianas y en la dipersión de distribuciones entre `+` y `-`, indicando una mayor relación con A15 y pontencial para predecir.

2. - **A8, A9 y A11:** presentan diferencias claras en las proporciones de `+` y `-`, mostrando una relación importante con A15.
- **A0, A3, A4, A5, A6 y A12:** presentan algunas diferencias entre sus categorías, aunque la relación es menos marcada.

3. Las variables con mayor potencial predictivo son **A2, A7, A10, A14, A8, A9 y A11**. Las variables **A1 y A13** parecen tener menor capacidad predictiva, mientras que **A0, A3, A4, A5, A6 y A12** presentan un potencial predictivo moderado.
