## Paso 4: Análisis de variables cuantitativas

Haga análisis de normalidad de las variables cuantitativas. Indique cuáles variables cumplen con el supuesto de normalidad y cuáles no.

Haga un análisis gráfico de outliers, y como resultado de este indique que variables tienen datos atípicos.

**Identificar variables cuantitativas**

```python
num_vars = df.select_dtypes(include=['number']).columns

num_vars
```

**Prueba de normalidad**

```python
from scipy.stats import kstest


for num in num_vars:
    _, p_value = kstest(
        df[num].dropna(),
        'norm',
        args=(df[num].mean(), df[num].std())
    )

    print(f'Prueba de normalidad para {num}')
    print('Valor p:', p_value)

    if p_value > 0.05:
        print('La variable cumple el supuesto de normalidad')
    else:
        print('La variable no cumple el supuesto de normalidad')

    print()
```

**Importar librería gráfica Y Distribución de variables cuantitativas (Histogramas)**

```python
import matplotlib.pyplot as plt
```

```python
import matplotlib.pyplot as plt

# Histogramas de variables cuantitativas

df[num_vars].hist(
    figsize=(16,8),
    bins=20,
    grid=False
)

plt.tight_layout()
plt.show()
```

**Prueba de normalidad Kolmogorov-Smirnov**

```python
from scipy.stats import kstest

# Prueba de normalidad para variables cuantitativas

for var in num_vars:
    estadistico, p_valor = kstest(
        df[var].dropna(),
        'norm'
    )

    print(f'Variable: {var}')
    print(f'Estadístico: {estadistico}')
    print(f'p-valor: {p_valor}')

    if p_valor > 0.05:
        print('Los datos podrían seguir una distribución normal')
    else:
        print('Los datos no siguen una distribución normal')

    print('-'*50)
```

**Análisis de valores atípicos (Outliers)**

```python
# Diagramas de caja para identificar valores atípicos

for var in num_vars:
    plt.figure(figsize=(8,3))

    df.boxplot(
        column=var,
        grid=False,
        vert=False
    )

    plt.title(f'Diagrama de caja de {var}')
    plt.show()
```

**Pregunta 1:**
Al aplicar la prueba de normalidad de Kolmogorov-Smirnov sobre las variables cuantitativas A1, A2, A7, A10, A13 y A14, se obtuvo que todas presentan un valor p menor a 0.05, por lo que se rechaza la hipótesis de normalidad. Esto indica que las variables analizadas no siguen una distribución normal y presentan diferencias respecto al comportamiento esperado de una distribución normal.

**Pregunta 2:**
A partir de los diagramas de caja realizados para las variables cuantitativas A1, A2, A7, A10, A13 y A14, se identificaron posibles valores atípicos en todas las variables analizadas. Algunas presentan pocos valores extremos, mientras que otras como A10, A13 y especialmente A14 muestran una mayor cantidad de datos alejados del comportamiento general. Estos valores deben ser evaluados posteriormente, ya que pueden representar casos reales dentro del conjunto de datos o influir en análisis posteriores.

**Relación entre variables cuantitativas y la variable objetivo A15**

```python
num_vars = df.select_dtypes(include=['number']).columns
```

```python
import matplotlib.pyplot as plt

for num in num_vars:
    plt.figure()
    df.boxplot(column=num, by='A15', grid=False, vert=False)
    plt.title(f'Relación entre {num} y A15')
    plt.suptitle('')
    plt.show()
```

**Relación entre variables cualitativas y A15**

```python
cat_vars = df.select_dtypes(include=['object']).columns.drop('A15')
```

```python
for var in cat_vars:
    tabla = pd.crosstab(df[var], df['A15'], normalize='index')
    tabla.plot.bar(stacked=True)
    plt.title(f'Relación entre {var} y A15')
    plt.ylabel('Proporción')
    plt.show()
```
