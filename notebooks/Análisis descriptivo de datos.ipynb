```python
from pathlib import Path

import pandas as pd

ROOT_DIR = Path().resolve().parent
DATA_DIR = ROOT_DIR / "data/raw"
file_path = 'Superstore.csv'

df = pd.read_csv(DATA_DIR / file_path, encoding='latin1')
df.head()
```

```python
df.info()
```

```python
# Convertir a datetime Order Date y Ship Date
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d-%m-%Y')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format='%d-%m-%Y')
df.info()
```

```python
df.head()
```

```python
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
df['Weekday'] = df['Order Date'].dt.weekday
df.head()
```

```python
df['Month name'] = df['Order Date'].dt.month_name()
df['Weekday name'] = df['Order Date'].dt.day_name()
df.info()
```

```python
df.head()
```

```python
df['Ship delay'] = (df['Ship Date'] - df['Order Date']).dt.days
df.info()
```

```python
df.head()
```

```python
df = df.drop(columns=['Row ID', 'Customer ID', 'Product ID'])
df.head()
```

```python
df = df.drop(columns=['Country', 'Month', 'Weekday'])
df.head()
```

```python
df['Quarter'] = df['Order Date'].dt.quarter
df.head()
```

```python
df.info()
```

```python
df.rename(columns={'Month name': 'Month', 'Weekday name': 'Weekday'}, inplace=True)
df.head()
```

```python
df = df.sort_values(by='Order Date').reset_index(drop=True)
df.head()
```

```python
# Reordenar columnas

df = df[['Order ID', 'Order Date', 'Ship Date', 'Ship delay', 'Year',
 'Quarter', 'Month', 'Weekday', 'Customer Name', 'Segment', 'City', 'State', 
 'Region', 'Product Name', 'Category', 'Sub-Category', 'Sales', 'Quantity', 'Discount', 
 'Profit']]

df.head() 
```

```python
processed_data_dir = ROOT_DIR / "data/processed"
df.to_csv(processed_data_dir / "Superstore_processed.csv", index=False, encoding='utf-8')
```
